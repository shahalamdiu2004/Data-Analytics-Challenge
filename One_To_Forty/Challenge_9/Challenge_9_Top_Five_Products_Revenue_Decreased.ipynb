```python
import pandas as pd

data= pd.read_csv("product_revenue.csv")

df= data.groupby(["product_name","year"])["revenue"].sum().unstack(fill_value=0).reset_index()
df.columns.name=None

df["revenue_diff"]= df[2023] - df[2022]
df["reveune_ratio"]= (df["revenue_diff"]/df[2022]).map("{:.2%}".format)

result= df[df["revenue_diff"]<0].sort_values("revenue_diff").head(5).reset_index(drop=True)

result
```
