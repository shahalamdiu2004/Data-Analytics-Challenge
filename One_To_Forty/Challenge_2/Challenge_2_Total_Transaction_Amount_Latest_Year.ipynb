```python
import pandas as pd
t= pd.read_csv("transaction.csv")
c= pd.read_csv("customers.csv")

df= t.merge(c, on="customer_id", how="left")
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
my= df["transaction_date"].dt.year.max()
f= df[df["transaction_date"].dt.year== my]

r= f.groupby("customer_name", as_index=False)["amount"].sum()[["customer_name","amount"]].rename(columns={"amount":"total_amount"})
#r= f.groupby("customer_name", as_index=False).sum(numeric_only=True)[["customer_name","amount"]].rename(columns={"amount":"total_amount"})

r
```
