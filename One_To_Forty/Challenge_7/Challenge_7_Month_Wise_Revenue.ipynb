```python
import pandas as pd

df= pd.read_csv("sales.csv")

df["order_date"]= pd.to_datetime(df["order_date"])

fl= df[df["order_date"].dt.year==2023].copy()

fl["month_no"]= fl["order_date"].dt.month
fl["month"]= fl["order_date"].dt.strftime("%b")
fl["revenue"]= fl["price_per_unit"] * fl["quantity"]

r= fl.groupby(["month_no","month"], as_index=False)["revenue"].sum().drop(columns="month_no")

r
```
