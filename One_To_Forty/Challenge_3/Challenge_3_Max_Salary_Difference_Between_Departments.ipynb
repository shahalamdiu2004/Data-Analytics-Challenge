```python
import pandas as pd

df= pd.read_csv("salaries.csv")

g= df.groupby("department")["salary"].max()

m= g["Marketing"]
e= g["Engineering"]

r= pd.DataFrame({"markering_max_salary":[m],
                 "engineering_max_salary":[e],
                 "max_salary_diff": [abs(m-e)]})
r
                 
```
