```python
import pandas as pd
p= pd.read_csv("products.csv")
s= pd.read_csv("suppliers.csv")

df= p.merge(s[["supplier_id","country"]], on ="supplier_id", how="left")

mp= df.groupby("country")["price"].agg(max_price="max")

r= df.merge(mp, on="country").query("price==max_price").sort_values("country").reset_index(drop=True)[["country","product_name","price"]]

r
```
